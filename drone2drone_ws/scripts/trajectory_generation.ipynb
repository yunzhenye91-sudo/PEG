```python
from math import cos, cosh, pi, sin
import time
import numpy as np
from matplotlib import pyplot
```

```python
def generate_s_trajectory(ini_obj_position):
	pos_traj = []
	cur_obj_position = ini_obj_position.copy()
	cur_obj_position[2] += 1.5
	
	#Takeoff and hover to wait for car
	for i in range(0, 500): 
		pos_traj.append(cur_obj_position.copy())

	sign = -1
	for i in range(0, 1001):
		cur_obj_position[1] = ini_obj_position[1] + (0.5 * sin(0.002*pi*i))
		if i % 1000 == 0:
			sign = -1*sign
		cur_obj_position[0] += 3*sign*(0.001)
		pos_traj.append(cur_obj_position.copy())

	for i in range(0, 500): 
		pos_traj.append(cur_obj_position.copy())
	return pos_traj
```

```python
ini_obj_position = [0.2,0.3,1.0]
pos_traj = generate_s_trajectory(ini_obj_position)
print(len(pos_traj))
pos_traj = np.asarray(pos_traj)
pyplot.plot(pos_traj[:,0], pos_traj[:,1])
```

```python
def generate_circle(ini_obj_position):
	pos_traj = []
	cur_obj_position = ini_obj_position.copy()
	cur_obj_position[2] += 1.5
	#Takeoff and hover to wait for car
	for i in range(0, 500): 
		pos_traj.append(cur_obj_position.copy())

	for i in range(0, 1500000): 
		cur_obj_position[0] = ini_obj_position[0] + (1 * sin(0.002*i))  #Omega = 0.001
		cur_obj_position[1] = ini_obj_position[1] + (1 * cos(0.002*i)) -1
		pos_traj.append(cur_obj_position.copy())
	return pos_traj
```

```python
ini_obj_position = [0.0,1.6,1.0]
pos_traj = generate_circle(ini_obj_position)
print(len(pos_traj))
pos_traj = np.asarray(pos_traj)
print(pos_traj[:,0])
pyplot.plot(pos_traj[:,0], pos_traj[:,1])
```
